```python
import numpy as np
from numba import njit, prange

@njit
def compute_g_column(p, alpha):
    beta = alpha / (alpha - 1)
    norma_beta = np.sum(np.abs(p)**beta)**(1.0 / beta)
    return p / (norma_beta + 1.0)

@njit
def finsler_metric_jit(p, q, f, g, alpha):
    diff = np.abs(q - p)
    term1 = np.sum(f * diff**alpha) ** (1.0 / alpha)
    term2 = np.sum(g * (f**(1.0/alpha)) * (q - p))
    return term1 + term2

@njit(parallel=True)
def finsler_matrix_numba(X, alpha):
    m, n = X.shape
    f_all = X**2                          # f_ref for all columns

    # Precompute g_all
    g_all = np.zeros((m, n))
    for i in prange(n):
        g_all[:, i] = compute_g_column(X[:, i], alpha)

    # Allocate distance matrix
    M = np.zeros((n, n))

    # Compute full matrix (parallel)
    for i in prange(n):
        p = X[:, i]
        f = f_all[:, i]
        g = g_all[:, i]

        for j in range(n):
            q = X[:, j]
            M[i, j] = finsler_metric_jit(p, q, f, g, alpha)

    return M

def compute_finsler_matrix_numba(df, alpha):
    X = df.to_numpy().astype(np.float64)
    ids = df.columns

    M = finsler_matrix_numba(X, alpha)

    M = np.triu(M) + np.triu(M, k=1).T

    return pd.DataFrame(M, index=ids, columns=ids)
```

```python
def robust_osi_fast(df, eps=1e-8):

    df = df.copy()
    df["angles"] = df["angles"].round(3)

    neuron_cols = df.columns.drop("angles")

    grouped = df.groupby("angles")[neuron_cols]

    mu = grouped.mean()
    sigma = grouped.std().fillna(0)

    # reliability weighting
    R_prime = mu / (sigma + eps)
    R_prime[R_prime < 0] = 0

    # ⚠️ angles already radians
    theta = mu.index.to_numpy()

    exp_term = np.exp(2j * theta)[:, None]

    R = R_prime.to_numpy()

    numerator = np.abs(np.sum(R * exp_term, axis=0))
    denominator = np.sum(R, axis=0) + eps

    osi = numerator / denominator

    return pd.DataFrame({
        "neuron": neuron_cols,
        "Robust OSI": osi
    })
```

```python
gratting_local_ratos = []
angles_gratting_local = []
for f in files:
    fullpath = os.path.join(folder, f)
    data = np.load(fullpath, allow_pickle=True).item()
    gratting_local_ratos.append(pd.DataFrame(data['sresp'].T))
    angles_gratting_local.append(pd.DataFrame(data['istim']))
    
```

```python
for i in range(len(gratting_local_ratos)):
    gratting_local_ratos[i]['angles'] = angles_gratting_local[i]
```

```python
gratting_local_ratos_ordered = []
for df in gratting_local_ratos:
    df_ = df.sort_values(by="angles").reset_index(drop=True)

    gratting_local_ratos_ordered.append(df_)
```

```python
column_osi_dfs_local = []

for df in gratting_local_ratos_ordered:
    theta = df['angles'].to_numpy()
    column_osi = []

    for col in df.columns[:-1]:
        r = df[col].to_numpy()
        osi = np.abs(np.sum(r * np.exp(2j * theta))) / np.sum(r)
        column_osi.append([col, osi])

    column_osi_df = pd.DataFrame(column_osi, columns=['neuron', 'OSI'])
    column_osi_df_ordered = column_osi_df.sort_values(by='OSI').reset_index(drop=True)

    column_osi_dfs_local.append(column_osi_df_ordered)
```

```python
column_robust_osi_dfs_local = []

for df in gratting_local_ratos_ordered:
    robust_osi_df = robust_osi_fast(df)

    column_osi_df_ordered = robust_osi_df.sort_values(by='Robust OSI').reset_index(drop=True)

    column_robust_osi_dfs_local.append(column_osi_df_ordered)
```

```python
merged_local = []
for df_1, df_2 in zip(column_osi_dfs_local,
                      column_robust_osi_dfs_local):
    merged_lis = df_1.merge(
        df_2,
        on="neuron",
        how="inner"
    )
    merged_local.append(merged_lis)
    
```

```python
betti_0_local = []
betti_1_local = []
persistent_betti_1_local = []

for df in local_resps:
    betti_0 = []
    betti_1 = []
    persistent_betti_1 = []

    for start, df_window in sliding_column_windows(df, 300):
        print(start)

        dist_M = compute_finsler_matrix_numba(df_window, 2)
        dist_M_np = dist_M.to_numpy()

        result = ripser(dist_M_np, distance_matrix=True, maxdim=1)
        diagrams = result['dgms']
        
        beta0 = diagrams[0]
        beta1 = diagrams[1]

        # total Betti-1
        betti_0.append(len(beta0))
        betti_1.append(len(beta1))

        # persistent Betti-1 ( > 10% of filtration )
        finite_beta1 = beta1[np.isfinite(beta1[:, 1])]

        if len(finite_beta1) > 0:
            persistence = finite_beta1[:, 1] - finite_beta1[:, 0]
            threshold = 0.1 * np.max(finite_beta1[:, 1])
            persistent_cycles = np.sum(persistence > threshold)
        else:
            persistent_cycles = 0

        persistent_betti_1.append(persistent_cycles)
    
    betti_0_local.append(betti_0)
    betti_1_local.append(betti_1)
    persistent_betti_1_local.append(persistent_betti_1)
```

```python
def sliding_column_windows(df, window_size, step=1):
    n_cols = df.shape[1]

    for start in range(0, n_cols - window_size + 1, step):
        yield start, df.iloc[:, start:start + window_size]
```

```python
gratting_noisy_ratos = []
angles_gratting_noisy = []
for f in files_noisy:
    fullpath = os.path.join(folder_noisy, f)
    
    # pula se for diretório
    if os.path.isdir(fullpath):
        continue
    
    data = np.load(fullpath, allow_pickle=True).item()
    gratting_noisy_ratos.append(pd.DataFrame(data['sresp'].T))
    angles_gratting_noisy.append(pd.DataFrame(data['istim']))

    
```

```python
for i in range(len(gratting_noisy_ratos)):
    gratting_noisy_ratos[i]['angles'] = angles_gratting_noisy[i]
```

```python
gratting_ratos_noisy_ordered = []
for df in gratting_noisy_ratos:
    df_ = df.sort_values(by="angles").reset_index(drop=True)
    df_ = df_.dropna(subset=["angles"])

    gratting_ratos_noisy_ordered.append(df_)
```

```python
column_osi_dfs_noisy = []

for df in gratting_ratos_noisy_ordered:
    theta = df['angles'].to_numpy()
    column_osi = []

    for col in df.columns[:-1]:
        r = df[col].to_numpy()
        osi = np.abs(np.sum(r * np.exp(2j * theta))) / np.sum(r)
        column_osi.append([col, osi])

    column_osi_df = pd.DataFrame(column_osi, columns=['neuron', 'OSI'])
    column_osi_df_ordered = column_osi_df.sort_values(by='OSI').reset_index(drop=True)

    column_osi_dfs_noisy.append(column_osi_df_ordered)
```

```python
column_robust_osi_dfs_noisy = []

for df in gratting_ratos_noisy_ordered:
    robust_osi_df = robust_osi_fast(df)

    column_osi_df_ordered = robust_osi_df.sort_values(by='Robust OSI').reset_index(drop=True)

    column_robust_osi_dfs_noisy.append(column_osi_df_ordered)
```

```python
merged_noisy = []
for df_1, df_2 in zip(column_osi_dfs_noisy,
                      column_robust_osi_dfs_noisy):
    merged_lis = df_1.merge(
        df_2,
        on="neuron",
        how="inner"
    )
    merged_noisy.append(merged_lis)
    
```

```python
plt.figure(figsize=(6,4))

# common bins (VERY important)
all_diff = np.concatenate([
    (df["Robust OSI"] - df["OSI"]).values
    for df in merged_noisy
])
bins = np.histogram_bin_edges(all_diff, bins=40)

for i, df in enumerate(merged_noisy):
    diff = df["Robust OSI"] - df["OSI"]

    plt.hist(
        diff,
        bins=bins,
        density=True,      # normalize
        alpha=0.35,
        histtype="stepfilled",
        edgecolor="black",
        linewidth=0.6,
        label=f"Mouse {i+1}"
    )

    # mean line (VERY informative)
    plt.axvline(diff.mean(), linestyle="--", linewidth=1)

plt.xlabel("Robust OSI − Classical OSI")
plt.ylabel("Probability density")
plt.legend(frameon=False)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
from scipy.stats import pearsonr

n = len(merged_noisy)

fig = plt.figure(figsize=(4*n + 1, 4))

gs = fig.add_gridspec(1, n + 1, width_ratios=[*([1]*n), 0.05])

axes = [fig.add_subplot(gs[0, i]) for i in range(n)]
cax = fig.add_subplot(gs[0, -1])

from scipy.stats import pearsonr

for i, (ax, df) in enumerate(zip(axes, merged_noisy)):

    x = df["OSI"]
    y = df["Robust OSI"]

    # correlation
    r, p = pearsonr(x, y)

    hb = ax.hexbin(
        x, y,
        gridsize=60,
        mincnt=1,
        bins='log'
    )

    ax.plot([0.9985, 1], [0.9985, 1], 'k--', linewidth=1)

    ax.set_title(f"Noisy - Mouse {i+1}")
    ax.set_xlabel("OSI")
    ax.set_aspect('equal')

    # ---- formatted p-value ----
    if p < 1e-3:
        p_text = f"{p:.1e}"   # scientific notation
    else:
        p_text = f"{p:.3f}"

    # ---- annotation ----
    ax.text(
        0.05, 0.95,
        f"r = {r:.3f}\np = {p_text}",
        transform=ax.transAxes,
        ha="left",
        va="top",
        fontsize=10,
        bbox=dict(facecolor="white", alpha=0.8, edgecolor="none")
    )


axes[0].set_ylabel("Vector OSI")

fig.colorbar(hb, cax=cax, label="Neuron count")

plt.show()
```

```python
def rolling_mean_np(lst, window=300):
    arr = np.asarray(lst)
    kernel = np.ones(window) / window
    return np.convolve(arr, kernel, mode="valid")

osi_dfs_noisy_mean = []
for df in column_osi_dfs_noisy:
    osi_dfs_noisy_mean.append(rolling_mean_np(df['OSI']))
```

```python
noisy_resps = []

for n in range(len(gratting_ratos_noisy_ordered)):
    df_ = gratting_ratos_noisy_ordered[n]
    df = column_osi_dfs_noisy[n]
    
    neuron_order = df['neuron'].tolist()
    neuron_order = [c for c in neuron_order if c in df_.columns[:-1]]

    sresp_df_ordered_osi = df_[neuron_order]
    
    noisy_resps.append(sresp_df_ordered_osi)
```

```python
# for i in range(len(noisy_resps)):
#     df = noisy_resps[i]
#     noisy_resps[i] = (df - df.mean()) / df.std()
```

```python

betti_1_noisy = []
persistent_betti_1_noisy = []

for df in noisy_resps:
    betti_1 = []
    persistent_betti_1 = []

    for start, df_window in sliding_column_windows(df, 300):
        print(start)

        dist_M = compute_finsler_matrix_numba(df_window, 2)
        dist_M_np = dist_M.to_numpy()

        result = ripser(dist_M_np, distance_matrix=True, maxdim=1)
        diagrams = result['dgms']
        
        beta1 = diagrams[1]

        # total Betti-1
        betti_1.append(len(beta1))

        # persistent Betti-1 ( > 10% of filtration )
        finite_beta1 = beta1[np.isfinite(beta1[:, 1])]

        if len(finite_beta1) > 0:
            persistence = finite_beta1[:, 1] - finite_beta1[:, 0]
            threshold = 0.1 * np.max(finite_beta1[:, 1])
            persistent_cycles = np.sum(persistence > threshold)
        else:
            persistent_cycles = 0

        persistent_betti_1.append(persistent_cycles)
    
    betti_1_noisy.append(betti_1)
    persistent_betti_1_noisy.append(persistent_betti_1)
```

```python
with open("betti_1_noisy.pkl", "wb") as f:
    pickle.dump(betti_1_noisy, f)
    
with open("persistent_betti_1_noisy.pkl", "wb") as f:
    pickle.dump(persistent_betti_1_noisy, f)
```
